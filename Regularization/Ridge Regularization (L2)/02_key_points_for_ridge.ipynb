# 5 key points for Ridge(L2) Regularization

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.datasets import load_diabetes, make_regression

from sklearn.preprocessing import PolynomialFeatures
from mlxtend.evaluate import bias_variance_decomp
```

## 1-- How the coefficients get affected by L2 Regularization?

```python
# Coefficients shrink towards 0 but never become equal to 0
```

```python
data = load_diabetes()
```

```python
df = pd.DataFrame(data.data,columns=data.feature_names)
df['TARGET'] = data.target
```

```python
df.head()
```

```python
df.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(data.data, data.target, test_size = 0.2, random_state=42)
```

```python
coefs = []
r2_scores = []

for i in [0,10,100,1000]:
    reg = Ridge(alpha=i)
    reg.fit(X_train,y_train)
    
    coefs.append(reg.coef_.tolist())
    y_pred = reg.predict(X_test)
    r2_scores.append(r2_score(y_test,y_pred))
```

```python
plt.figure(figsize=(14,9))
plt.subplot(221)
plt.bar(data.feature_names,coefs[0])
plt.title('Alpha = 0 ,r2_score = {}'.format(round(r2_scores[0],2)))

plt.subplot(222)
plt.bar(data.feature_names,coefs[1])
plt.title('Alpha = 10 ,r2_score = {}'.format(round(r2_scores[1],2)))

plt.subplot(223)
plt.bar(data.feature_names,coefs[2])
plt.title('Alpha = 100 ,r2_score = {}'.format(round(r2_scores[2],2)))

plt.subplot(224)
plt.bar(data.feature_names,coefs[3])
plt.title('Alpha = 1000 ,r2_score = {}'.format(round(r2_scores[3],2)))

plt.show()
```

```python
# The range of values of coeffs shrink from -1000-750 to -0.4-0.8 as we increase alpha [coeffs shrink towards 0]
```

## 3 -- How is bias and variance affected by ridge regression?

```python
np.random.seed(42)
m = 100
X = 5 * np.random.rand(m, 1) - 2
y = 0.7 * X ** 2 - 2 * X + 3 + np.random.randn(m, 1)

plt.scatter(X, y)
plt.show()
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y.ravel(),test_size=0.3,random_state=42)
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

# Scale the polynomial features so alpha penalises all of them on a comparable scale
def make_model(alpha):
    return make_pipeline(PolynomialFeatures(degree=15), StandardScaler(), Ridge(alpha=alpha))
```

```python
alphas = np.logspace(-4, 2, 40)

loss = []
bias = []
variance = []

for i in alphas:
    avg_expected_loss, avg_bias, avg_var = bias_variance_decomp(
        make_model(i), X_train, y_train, X_test, y_test,
        loss='mse',
        num_rounds=200,
        random_seed=123)
    loss.append(avg_expected_loss)
    bias.append(avg_bias)
    variance.append(avg_var)
```

```python
# MLxtend library is used above!
```

```python
plt.plot(alphas,loss,label='Loss (bias + variance)')
plt.plot(alphas,bias,label='Bias')
plt.plot(alphas,variance,label='Variance')
plt.xscale('log')
plt.yscale('log')   # variance is huge at tiny alpha, so a linear axis hides everything else
plt.xlabel('Alpha')
plt.legend()
plt.show()
```

```python
# As alpha is increased, bias increases and variance decreases. We need to find sweet stop where bias <= variance(bias almost equal to var)
```

# 4-- How is loss function affected by ridge regression?

## It shrinks and moves towards origin

```python
X,y = make_regression(n_samples=100, n_features=1, n_informative=1, n_targets=1,noise=20,random_state=13)

plt.scatter(X,y)

from sklearn.linear_model import LinearRegression

reg = LinearRegression()
reg.fit(X,y)
print(reg.coef_)
print(reg.intercept_)
```

```python
def predict(m):
    return m*X - 2.29
```

```python
def cal_loss(m,alpha):
    return np.sum((y - m*X.ravel() + 2.29)**2) + alpha*m*m
```

```python
m = np.linspace(-45,100,100)
plt.figure(figsize=(4,6))
for j in [0,10,20,30,40,50,100]:
    loss = []
    for i in range(m.shape[0]):
        loss_i = cal_loss(m[i],j)
        loss.append(loss_i)
    plt.plot(m,loss,label='alpha = {}'.format(j))
plt.legend()
plt.xlabel('Alpha')
plt.ylabel('Loss')
plt.show()
```
